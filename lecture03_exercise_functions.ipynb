```python
# define book class (structure for containing book information).
class Book:
    def __init__(self, title, author, year, genre, country):
        self.title = title
        self.author = author
        self.year = year
        self.genre = genre
        self.country = country

```

```python
# 1. Create a dictionary with 20 objects of the Book class. 

books_db = {"1984": Book ("1984", "George Orwell", 1949, "Dystopian", "UK"),
            "To Kill a Mockingbird": Book("To Kill a Mockingbird", "Harper Lee", 1960, "Fiction", "USA"),
            "Pride and prejudice": Book( "Pride and prejudice", "Jane Austin", 1813, "Novel", "England")
           }

print(len(books_db))
print(books_db["1984"].author)

print(books_db["Pride and prejudice"].year)
```

```python
# 2. Define a function that find_books_by_genre(genre) that allows for searching the dictionary by genre.
def find_books_by_genre (genre):
    result = []
    for book in books_db.values():
        if book.genre == genre:
            result.append(book)
        return result

Dystopian_books = find_books_by_genre("Dystopian")
for book in Dystopian_books: 
    print(book.title)

print(Dystopian_books)

```

```python
# 3. Define a function that count_books_before_year(year) that allows for counting the number of books 
# written before a specific year. 

def count_books_before_year(year):
    count = 0
    for book in books_db.values():
        if book.year < year:
            count += 1
    return count

antal = count_books_before_year(1951)
print(f"Antal bøger før 1950: {antal}")
        
```
