## 6. Reflection
Variables help us organize, access, and analyze cultural data efficiently!

Add a markdown chunk below where you summarize key features of the different types of variables you have learned about today:

- Regular string
- Regular integer
- Regular floating point
- Container variables:
    - List
    - Set
    - Tuple
    - Dictionary

Strings: Stores text, such as names, titles, or genres. Strings are written inside quotation marks, ex."Iris".
	
Integer: Stores whole numbers without decimals, such as a year or an age, ex., 1998.

Floating point: Stores numbers with decimals, such as ratings or prices. Ex, 9.5.

List: Stores multiple items in a specific order (ordered). Lists can be changed, and they can contain duplicate items. ex, ["Milk", "Eggs", "Bread", ]

Set: Stores multiple unique items. Sets do not allow duplicates and do not have a fixed order (unordered). Items can be added or removed using methods such as .add().

Tuple: Stores multiple items in a specific order. A tuple cannot be changed after it is created.

Dictionary: Stores data as key-value pairs. It is useful for organising information about something, such as a song. Ex, {"artist": "John Lennon", "year": 1971}.
